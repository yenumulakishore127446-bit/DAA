### Python Implementations of Sorting Algorithms

Below are the Python equivalents of the sorting algorithms provided in the C++ code, along with a benchmark to measure their execution times.

```python
# Main execution for benchmarking

def run_benchmark(n):
    print(f"Number of Elements = {n}\n")

    arr = [random.randint(0, 1000) for _ in range(n)]

    # Bubble Sort
    temp_arr = list(arr)
    start_time = time.perf_counter_ns() # Using nanoseconds for precision
    bubble_sort(temp_arr)
    end_time = time.perf_counter_ns()
    print(f"Bubble Sort Time    : {(end_time - start_time) / 1000:.3f} microseconds")

    # Selection Sort
    temp_arr = list(arr)
    start_time = time.perf_counter_ns()
    selection_sort(temp_arr)
    end_time = time.perf_counter_ns()
    print(f"Selection Sort Time : {(end_time - start_time) / 1000:.3f} microseconds")

    # Insertion Sort
    temp_arr = list(arr)
    start_time = time.perf_counter_ns()
    insertion_sort(temp_arr)
    end_time = time.perf_counter_ns()
    print(f"Insertion Sort Time : {(end_time - start_time) / 1000:.3f} microseconds")

    # Merge Sort
    temp_arr = list(arr)
    start_time = time.perf_counter_ns()
    merge_sort(temp_arr)
    end_time = time.perf_counter_ns()
    print(f"Merge Sort Time     : {(end_time - start_time) / 1000:.3f} microseconds")

    # Quick Sort
    temp_arr = list(arr)
    start_time = time.perf_counter_ns()
    quick_sort(temp_arr)
    end_time = time.perf_counter_ns()
    print(f"Quick Sort Time     : {(end_time - start_time) / 1000:.3f} microseconds")

# Run the benchmark with n=100 as in the C++ example
run_benchmark(100)
```
